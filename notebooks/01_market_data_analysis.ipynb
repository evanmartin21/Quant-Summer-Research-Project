## 6. Calculate Returns

We can calculate daily percentage returns:

$$
R_t = \frac{P_t-P_{t-1}}{P_{t-1}}
$$

```python
spy["Simple_Return"] = spy["Close"].pct_change() 

# Introducing a new column to compute daily percentage returns

spy.head()
```

The daily logarithmic return is calculated as:

$$
r_t = \ln\left(\frac{P_t}{P_{t-1}}\right)
$$

where:

- $r_t$ is the logarithmic return at time $t$
- $P_t$ is the closing price at time $t$
- $P_{t-1}$ is the closing price on the previous trading day

```python
spy["Log_Return"] = np.log(spy["Close"] / spy["Close"].shift(1))

# Introducing a new column to compute daily logarithmic return

spy.head()
```

This causes some issues because the first entry of both new columns is 'NaN' which we can sort. Let us find the missing values and delete them.

```python
spy.isnull().sum()
```

```python
spy = spy.dropna()

spy.head()
```

## 7. Analyse Return Distributions

### 7.1 Summary Statistics

We now look to calculate the distributional properties of SPY returns, including average returns, volatility, and any extreme observations (outliers).

```python
spy["Log_Return"].describe()
```

The mean return is calculated as:

$\mu = \frac{1}{n}\sum_{t=1}^{n} r_t$

where $r_t$ represents the logarithmic return at time $t$ and $n$ represents the number of observations.

The standard deviation of returns is used as a measure of volatility:

$\sigma = \sqrt{\frac{1}{n-1}\sum_{t=1}^{n}(r_t-\mu)^2}$

where $\sigma$ represents the standard deviation of returns.

### 7.2 Annualised Statistics

We can use this to approximate annualised returns and volatility:

$\mu_{annual}=252\mu_{daily}$

$\sigma_{annual}=\sigma_{daily}\sqrt{252}$

```python
daily_mean = spy["Log_Return"].mean()
daily_volatility = spy["Log_Return"].std()

annual_return = daily_mean * 252
annual_volatility = daily_volatility * np.sqrt(252)

print("Annualised Return:", annual_return)
print("Annualised Volatility:", annual_volatility)
```

### 7.3 Plots

The data seems to be concentrated around zero, showing that most daily price movements are small; equally there is a possible negative skew, which is a common observation in equity markets. We can test the data to see if this is true.

### 7.3 Measuring Skewness

Skewness measures the asymmetry of a distribution:

$$
Skewness = \frac{E[(r_t - \mu)^3]}{\sigma ^3}
$$

where:
- $\mu$ = mean return
- $\sigma$ = standard deviation
- $E$ = expected value

```python
spy["Log_Return"].skew()
```

The output shows a negative skew, confirming the fact that the left-tail is longer, which means that the extreme returns are more likely to be negative. 

Perhaps it may also be beneficial to compute the kurtosis: 

$$
Kurtosis = \frac{E[(r_t - \mu)^4]}{\sigma ^4}
$$

```python
spy["Log_Return"].kurtosis()
```

This is an abnormally large reading for a normal distribution, which highlights that SPY daily log returns exhibit very heavy tails. This could cause issues later when we try to use models such as Black-Scholes, which assumes that daily log returns follow a normal distribution. 

## 8. Calculating Historical Volatility

We will now look at the rolling volatility over a window of $n$ days:

$$
\sigma_t = \sqrt{\frac{1}{n-1}\sum_{i=t-n+1}^{t}(r_i-\bar{r}_t)^2}
$$

where:

- $\sigma_t$ = volatility at time $t$
- $r_i$ = individual log returns
- $\bar{r}$ = average return over the rolling window
- $n$ = window size

We can pick a window of $n=30$ to calculate the rolling volatility across the previous 30 trading days.

```python
rolling_volatility = spy["Log_Return"].rolling(window=30).std()

rolling_volatility.head(35)
```

This is unordinary, we see 29 entries 'NaN' because we are asking the code to calculate the volatility across the previous 30 daily log returns. 

Here most values range from $0.008215-0.009982$, showing that on any one day, SPY prices were moving by $<1%$ per day over a 30 day period. Extrapolating for annualised volatility: $0.009981$ x $\sqrt{252} \approx 0.158$ which is expected of SPY.

```python
spy["Rolling_Volatility"] = spy["Log_Return"].rolling(window=30).std() * np.sqrt(252)

# Introducing a new column to monitor the rolling volatility over a 30 day period.

spy.iloc[29:34] 
```

The reason for mentioning this is again due to the Black-Scholes model assuming constant volatility, motivating the possibility to explore more sophisticated models.
